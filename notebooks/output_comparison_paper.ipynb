<h1>Output Comparison</h1>


The following file can be used to compare any simulation runs no matter which settings have been used. To do so, one needs to provide the unique part of file path for each run in the file_extensions list. The notebook will then load the data and display the results for comparison for the four indicators:

1. Required Beds
2. Occupancy
3. Proportion rejected
4. Weekly Capacity Transfers

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
import sys
from itertools import product
from scipy.stats import ttest_rel, norm, sem, t

pd.set_option("display.max_colwidth", None)

sns.set(style="whitegrid", palette="colorblind", font_scale=1.2)
sns.set_context("paper")
# Add the project folder to the Python path
project_path = os.path.abspath(os.path.join(".."))
if project_path not in sys.path:
    sys.path.append(project_path)
from model.config.run_config import *

import warnings

warnings.filterwarnings("ignore")
```

```python
# list of file names
base_file_name = "admission_log"
output_folder = "../model/outputs"

file_extensions = [
    "_hospitalScenario_nicuWaitTimeLever_nicuWaitTimeLeverChange_2_runs_100.csv",
    "_hospitalScenario_phototherapyIntervention_nicuWaitTimeLever_nicuWaitTimeLeverChange_2_runs_100.csv",
    "_hospitalScenario_sepsisIntervention_nicuWaitTimeLever_nicuWaitTimeLeverChange_2_runs_100.csv",
    "_hospitalScenario_nicuWaitTimeLever_nicuWaitTimeLeverChange_2_predictionModelIntervention_predictionModelWeekThresholdChange_-2_runs_100.csv",
    "_hospitalScenario_phototherapyIntervention_sepsisIntervention_nicuWaitTimeLever_nicuWaitTimeLeverChange_2_predictionModelIntervention_predictionModelWeekThresholdChange_-2_runs_100.csv",
]

WARM_UP_TIME = 70
```

```python
def read_files(base_file_name, use_warm_up_time=True):
    data = []
    for extension in file_extensions:
        # Construct the full file path
        file_path = f"{output_folder}/{base_file_name}{extension}"
        print(f"Reading file: {file_path}")
        # Read the CSV file
        df = pd.read_csv(file_path)
        if use_warm_up_time:
            # Filter rows based on WARM_UP_TIME
            df = df[df["start_date"] >= WARM_UP_TIME]
            # Adjust start_date by subtracting warm-up time
            df["start_date"] = df["start_date"] - WARM_UP_TIME

        # Append the filtered DataFrame to the data list
        data.append(df)

    return data


def calculate_confidence_interval(data, confidence=0.95):
    n = len(data)
    if n < 2:
        mean_val = np.mean(data)
        return mean_val, mean_val, mean_val
    mean = np.mean(data)
    se = sem(data)  # Use scipy's sem function
    h = se * t.ppf((1 + confidence) / 2.0, df=n - 1)  # Use t-distribution
    return mean, mean - h, mean + h
```

<h2>Required Bed Count</h2>


```python
def calculate_percentages(df):
    df_inside = df[df["subregion"] != "Outside Region"]
    result = df_inside.groupby(["run_id", "ward"])["length_of_stay"].sum().reset_index()
    occupancy_rate_aim = 0.85
    result["used"] = (result["length_of_stay"] / 365) / (occupancy_rate_aim)
    return result


def process_data_frames(data_frames):
    summary_stats = []
    ttest_results = []
    reference_df = None

    for i, df in enumerate(data_frames):
        percentages = calculate_percentages(df)

        summary = {"DataFrame": i}
        for ward in percentages["ward"].unique():
            ward_data = percentages[percentages["ward"] == ward]["used"].values
            mean, ci_lower, ci_upper = calculate_confidence_interval(ward_data)
            summary[f"{ward}_mean"] = mean
            summary[f"{ward}_CI_lower"] = ci_lower
            summary[f"{ward}_CI_upper"] = ci_upper

        summary_stats.append(summary)

        if i == 0:
            reference_df = percentages
        else:
            ttest_summary = {"DataFrame": f"Comparison with DataFrame 0 and {i}"}
            for ward in percentages["ward"].unique():
                if ward in reference_df["ward"].unique():
                    used_0 = reference_df[reference_df["ward"] == ward]["used"].values
                    used_i = percentages[percentages["ward"] == ward]["used"].values
                    t_stat, p_value = ttest_rel(used_0, used_i)
                    ttest_summary[f"{ward}_t_stat"] = t_stat
                    ttest_summary[f"{ward}_p_value"] = p_value
            ttest_results.append(ttest_summary)

    summary_table = pd.DataFrame(summary_stats)
    ttest_table = pd.DataFrame(ttest_results)

    return summary_table, ttest_table


def format_results(summary_table, ttest_table):
    formatted_results = []

    for i, row in summary_table.iterrows():
        formatted_row = {"DataFrame": row["DataFrame"]}
        for col in summary_table.columns:
            if "_mean" in col:
                ward = col.replace("_mean", "")

                # Round to 1 decimal place
                mean = round(row[col], 1)
                lower_ci = round(row[f"{ward}_CI_lower"], 1)
                upper_ci = round(row[f"{ward}_CI_upper"], 1)

                confidence_interval = f"({lower_ci}-{upper_ci})"

                star = ""
                p_value_text = ""

                # Check for significance and add p-value
                if i > 0:
                    p_value = ttest_table.loc[
                        ttest_table["DataFrame"]
                        == f"Comparison with DataFrame 0 and {i}",
                        f"{ward}_p_value",
                    ]
                    if not p_value.empty:
                        p_value = p_value.values[0]
                        p_value_text = f", p={p_value:.3f}"
                        if p_value < 0.05:
                            star = "*"  # Add star after the mean if significant
                        if p_value < 0.001:
                            star = "**"

                # Combine mean, star, confidence interval, and p-value
                formatted_row[ward] = (
                    f"{mean}{star} {confidence_interval}{p_value_text}"
                )

        formatted_results.append(formatted_row)

    formatted_df = pd.DataFrame(formatted_results).set_index("DataFrame")
    return formatted_df


# Process admissions data
summary_table, ttest_table = process_data_frames(admissions)
formatted_df = format_results(summary_table, ttest_table)
formatted_df
```

<h2>Proportion rejected </h2>


```python
def calculate_percentages(df, capacity_transfers):
    results = []
    for run in capacity_transfers["run_id"].unique():
        run_results = {"run_id": run}
        for ward in capacity_transfers[
            capacity_transfers["subregion"] != "Outside Region"
        ]["ward"].unique():
            ward_df = df[(df["ward"] == ward) & (df["run_id"] == run)]
            if len(ward_df) > 0:
                ward_group = (
                    capacity_transfers[capacity_transfers["run_id"] == run]
                    .groupby("ward")
                    .size()
                )
                if ward in ward_group:
                    percentage = ward_group[ward] / len(ward_df) * 100
                else:
                    percentage = 0
            else:
                percentage = 0
            run_results[ward] = percentage
        results.append(run_results)
    return pd.DataFrame(results)


def process_data_frames(data_frames):
    summary_stats = []
    ttest_results = []

    reference_df = None

    for i, df in enumerate(data_frames):
        df_outside = df[df["hospital"] == "Outside Hospital"]
        capacity_transfers = df_outside[df_outside["subregion"] != "Outside Region"]
        results_df = calculate_percentages(df, capacity_transfers)

        summary = {"DataFrame": i}
        for ward in results_df.columns:
            if ward != "run_id":
                mean, ci_lower, ci_upper = calculate_confidence_interval(
                    results_df[ward].dropna()
                )
                summary[f"{ward}_mean"] = mean
                summary[f"{ward}_CI_lower"] = ci_lower
                summary[f"{ward}_CI_upper"] = ci_upper

        summary_stats.append(summary)

        if i == 0:
            reference_df = results_df
        else:
            ttest_summary = {"DataFrame": f"Comparison with DataFrame 0 and {i}"}
            for ward in results_df.columns:
                if ward != "run_id" and ward in reference_df.columns:
                    t_stat, p_value = ttest_rel(
                        reference_df[ward].dropna(), results_df[ward].dropna()
                    )
                    ttest_summary[f"{ward}_t_stat"] = t_stat
                    ttest_summary[f"{ward}_p_value"] = p_value
            ttest_results.append(ttest_summary)

    summary_table = pd.DataFrame(summary_stats)
    ttest_table = pd.DataFrame(ttest_results)

    return summary_table, ttest_table


def format_results(summary_table, ttest_table):
    formatted_results = []

    for i, row in summary_table.iterrows():
        formatted_row = {"DataFrame": row["DataFrame"]}
        for col in summary_table.columns:
            if "_mean" in col:
                ward = col.replace("_mean", "")

                # Check if the ward is "medium ward" for 2 decimal rounding, else 1 decimal
                if ward == "medium":
                    mean = round(row[col], 2)
                    lower_ci = round(row[f"{ward}_CI_lower"], 2)
                    upper_ci = round(row[f"{ward}_CI_upper"], 2)
                else:
                    mean = round(row[col], 1)
                    lower_ci = round(row[f"{ward}_CI_lower"], 1)
                    upper_ci = round(row[f"{ward}_CI_upper"], 1)

                confidence_interval = f"({lower_ci}-{upper_ci})"

                star = ""
                p_value_text = ""

                # Check for significance and add p-value
                if i > 0:
                    p_value = ttest_table.loc[
                        ttest_table["DataFrame"]
                        == f"Comparison with DataFrame 0 and {i}",
                        f"{ward}_p_value",
                    ]
                    if not p_value.empty:
                        p_value = p_value.values[0]
                        p_value_text = f", p={p_value:.3f}"
                        if p_value < 0.05:
                            star = "*"  # Add star after the mean if significant
                        if p_value < 0.001:
                            star = "**"

                # Combine mean, star, confidence interval, and p-value
                formatted_row[ward] = (
                    f"{mean}{star} {confidence_interval}{p_value_text}"
                )

        formatted_results.append(formatted_row)
    formatted_df = pd.DataFrame(formatted_results).set_index("DataFrame")
    return formatted_df


# Assuming `admissions` is a list of DataFrames
summary_table, ttest_table = process_data_frames(admissions)
formatted_df = format_results(summary_table, ttest_table)
```

<h2>Occupancy </h2>


```python
base_file_name = "occupancy_by_hospital"
occupancy_rates = read_files(base_file_name)
```

```python
def process_occupancy_data(df_occupancy):
    df_occupancy["start_date"] = df_occupancy["start_date"].astype(float)
    df_occupancy["moving_average"] = df_occupancy.groupby(["hospital", "run_id"])[
        "occupancy"
    ].transform(lambda x: x.rolling(window=7, min_periods=1).mean())
    df_occupancy["moving_std"] = df_occupancy.groupby(["hospital", "run_id"])[
        "occupancy"
    ].transform(lambda x: x.rolling(window=7, min_periods=1).std())
    return df_occupancy


def calculate_confidence_interval(data, confidence=0.95):
    """
    data: 1-D array of per-run mean occupancies (fractions).
    returns: (mean%, lower_CI%, upper_CI%).
    """
    n = len(data)
    if n < 2:
        mean = np.mean(data)
        return mean * 100, mean * 100, mean * 100
    mean = np.mean(data)
    se = sem(data)  # Use scipy's sem function
    h = se * t.ppf((1 + confidence) / 2.0, df=n - 1)  # Use t-distribution
    return mean * 100, (mean - h) * 100, (mean + h) * 100


def compare_occupancy_by_ward_level(data_frames):
    comparison_stats = []
    ttest_results = []
    reference_means = None

    for i, df in enumerate(data_frames):
        # preprocess
        df_occ = process_occupancy_data(df)

        # 1) per-run means
        run_means = (
            df_occ.groupby(["ward_level", "run_id"])["occupancy"]
            .mean()
            .reset_index(name="run_mean")
        )

        # 2) mean ± CI across runs
        ward_stats = (
            run_means.groupby("ward_level")["run_mean"]
            .apply(lambda x: calculate_confidence_interval(x.dropna()))
            .reset_index(name="ci_tuple")
        )
        ward_stats[["mean", "CI_Lower", "CI_Upper"]] = pd.DataFrame(
            ward_stats["ci_tuple"].tolist(), index=ward_stats.index
        )
        ward_stats.drop(columns=["ci_tuple"], inplace=True)

        # 3) min & max across runs
        min_max = (
            run_means.groupby("ward_level")["run_mean"]
            .agg(min="min", max="max")
            .reset_index()
        )
        # convert to %
        min_max[["min", "max"]] *= 100
        ward_stats = ward_stats.merge(min_max, on="ward_level")

        ward_stats["file"] = f"DataFrame_{i}"
        comparison_stats.append(ward_stats)

        # 4) paired t-test vs first DF
        if i == 0:
            reference_means = run_means
        else:
            ttest_summary = {"comparison_with_file_0": i}
            for wl in ward_stats["ward_level"]:
                merged = pd.merge(
                    reference_means[reference_means["ward_level"] == wl],
                    run_means[run_means["ward_level"] == wl],
                    on="run_id",
                    suffixes=("_0", "_i"),
                )
                if len(merged) > 1:
                    _, p_value = ttest_rel(
                        merged["run_mean_0"] * 100, merged["run_mean_i"] * 100
                    )
                    ttest_summary[f"{wl}_p_value"] = p_value
            ttest_results.append(ttest_summary)

    # 5) pivot to wide form including min & max
    comparison_table = pd.concat(comparison_stats, ignore_index=True)
    pivot_table = comparison_table.pivot(
        index="file",
        columns="ward_level",
        values=["mean", "CI_Lower", "CI_Upper"],
    )
    pivot_table.columns = [f"{col[1]}_{col[0]}" for col in pivot_table.columns.values]

    ttest_result_df = pd.DataFrame(ttest_results)
    return pivot_table, ttest_result_df


def format_results(pivot_table, ttest_result_df):
    formatted_results = []

    for i in range(len(ttest_result_df) + 1):
        row = {"file": f"DataFrame_{i}"}
        for ward_level in set(col.split("_")[0] for col in pivot_table.columns):
            mean = pivot_table.get(f"{ward_level}_mean", pd.Series([None])).iloc[i]
            lo, hi = (
                pivot_table.get(f"{ward_level}_CI_Lower", pd.Series([None])).iloc[i],
                pivot_table.get(f"{ward_level}_CI_Upper", pd.Series([None])).iloc[i],
            )

            if pd.notna(mean):
                # rounding
                dp = 2 if ward_level == "medium" else 1
                mean, lo, hi = (
                    round(mean, dp),
                    round(lo, dp),
                    round(hi, dp),
                )

                ci_str = f"({lo}-{hi})"
                star = ""
                p_text = ""

                if i > 0:
                    p_series = ttest_result_df.loc[
                        ttest_result_df["comparison_with_file_0"] == i,
                        f"{ward_level}_p_value",
                    ]
                    if not p_series.empty:
                        p = p_series.values[0]
                        p_text = f", p={p:.3f}"
                        if p < 0.05:
                            star = "*"
                        if p < 0.001:
                            star = "**"

                # now include min–max after CI & p-value
                row[ward_level] = f"{mean}{star} {ci_str}{p_text}"
            else:
                row[ward_level] = "N/A"

        formatted_results.append(row)

    return pd.DataFrame(formatted_results).set_index("file")


comparison_table, ttest_results = compare_occupancy_by_ward_level(occupancy_rates)
formatted_df = format_results(comparison_table, ttest_results)
formatted_df
```

<h2>Weekly Transfers</h2>


```python
base_file_name = "occupancy_by_hospital"
occupancy_rates = read_files(base_file_name, False)
base_file_name = "admission_log"
admissions = read_files(base_file_name, False)
```

```python
def process_occupancy_data(df_occupancy, warm_up_time):
    # Convert start_date to float and adjust by subtracting warm-up time
    df_occupancy["start_date"] = df_occupancy["start_date"].astype(float)
    df_occupancy["date"] = df_occupancy["start_date"] - warm_up_time

    # Calculate moving average per hospital and run
    df_occupancy["moving_average"] = df_occupancy.groupby(["hospital", "run_id"])[
        "occupancy"
    ].transform(lambda x: x.rolling(window=7, min_periods=1).mean())

    return df_occupancy


def merge_and_calculate(data_frames, occupancy_rates, warm_up_time):
    all_weekly_transfer = []

    for i, (df_transfers, df_occupancy) in enumerate(zip(data_frames, occupancy_rates)):
        # Process occupancy data
        df_occupancy = process_occupancy_data(df_occupancy, warm_up_time)

        # Aggregate transfer data
        transfer_agg = (
            df_transfers.groupby(["run_id", "start_date", "ward"])
            .size()
            .reset_index(name="num_transfers")
        )

        # Rename columns for merging
        transfer_agg.rename(
            columns={"ward": "ward_level", "start_date": "date"}, inplace=True
        )

        # Aggregate occupancy data
        occupancy_agg = df_occupancy.groupby(
            ["run_id", "date", "ward_level"], as_index=False
        ).agg({"moving_average": "mean"})

        # Merge occupancy and transfer data
        merged_data = pd.merge(
            occupancy_agg, transfer_agg, how="left", on=["run_id", "date", "ward_level"]
        ).fillna({"num_transfers": 0})

        # Ensure 'num_transfers' is an integer
        merged_data["num_transfers"] = merged_data["num_transfers"].astype(int)

        # Calculate rolling sum of transfers
        merged_data["transfer_count"] = (
            merged_data.groupby(["run_id", "ward_level"])["num_transfers"]
            .transform(lambda x: x.rolling(window=7, min_periods=1).sum())
            .fillna(0)
        )

        # Round occupancy rates for analysis
        factor = 25
        merged_data["moving_average_rounded"] = (
            (merged_data["moving_average"] * factor).round() / factor * 100
        )

        # Calculate weekly transfer for each occupancy rate and ward level
        weekly_transfer = (
            merged_data.groupby(["run_id", "ward_level", "moving_average_rounded"])
            .agg(
                weekly_transfer=("transfer_count", "mean"),
                sample_size=("transfer_count", "size"),
            )
            .reset_index()
        )

        # Filter out small sample sizes
        MIN_SAMPLE_SIZE = 8
        weekly_transfer = weekly_transfer[
            weekly_transfer["sample_size"] >= MIN_SAMPLE_SIZE
        ].drop(columns=["sample_size"])

        # Add file identifier
        weekly_transfer["file"] = i
        all_weekly_transfer.append(weekly_transfer)

    # Concatenate all weekly transfer data
    final_weekly_transfer = pd.concat(all_weekly_transfer, ignore_index=True)
    return final_weekly_transfer


def mean_confidence_interval(data, confidence=0.95):
    n = len(data)
    mean = np.mean(data)
    se = sem(data)
    h = se * t.ppf((1 + confidence) / 2.0, n - 1)
    return mean, mean - h, mean + h


# Calculate aggregated data
def calculate_aggregated_data(final_weekly_transfer):
    aggregated_data = (
        final_weekly_transfer.groupby(["file", "ward_level", "moving_average_rounded"])
        .agg(
            weekly_transfer_mean=("weekly_transfer", np.mean),
            weekly_transfer_ci_lower=(
                "weekly_transfer",
                lambda x: mean_confidence_interval(x)[1],
            ),
            weekly_transfer_ci_upper=(
                "weekly_transfer",
                lambda x: mean_confidence_interval(x)[2],
            ),
        )
        .reset_index()
    )
    return aggregated_data


def plot_weekly_transfer(aggregated_data):
    # Define labels for each scenario
    labels = ["Base", "Sepsis", "Phototherapy", "Change NICU guidelines", "Combined"]
    # Mapping for better titles
    ward_title_map = {"NICU": "NICU", "high": "HCN", "medium": "MCN"}

    # Unique ward levels
    ward_levels = aggregated_data["ward_level"].unique()

    # Create a color palette and map for files
    palette = sns.color_palette("tab10", aggregated_data["file"].nunique())
    color_map = {
        file_num: color
        for file_num, color in zip(aggregated_data["file"].unique(), palette)
    }

    # Plot separate graphs for each ward level
    for ward_level in ward_levels:
        subset = aggregated_data[aggregated_data["ward_level"] == ward_level]
        plt.figure(figsize=(10, 6), dpi=400)

        for file_num in subset["file"].unique():
            file_subset = subset[subset["file"] == file_num]
            plt.plot(
                file_subset["moving_average_rounded"],
                file_subset["weekly_transfer_mean"],
                marker="o",
                markersize=4,
                label=labels[file_num],
                color=color_map[file_num],
            )

            # Plot confidence intervals
            plt.fill_between(
                file_subset["moving_average_rounded"],
                file_subset["weekly_transfer_ci_lower"],
                file_subset["weekly_transfer_ci_upper"],
                color=color_map[file_num],
                alpha=0.2,
            )

        plt.title(f"{ward_title_map.get(ward_level, ward_level)} Ward Level")

        plt.xlabel("Weekly Occupancy (%)")
        plt.ylabel("Average Weekly Rejections")
        plt.legend()
        plt.grid(True, linestyle="--", alpha=0.7)
        plt.tight_layout()
        plt.show()


# Execute functions
def filter_data_frames(data_frames):
    results = []
    for df in data_frames:
        df_outside = df[df["hospital"] == "Outside Hospital"]
        capacity_transfers = df_outside[df_outside["subregion"] != "Outside Region"]
        results.append(capacity_transfers)
    return results


def calculate_aggregated_data(final_weekly_transfer):
    # Step 1: Aggregate by run first
    per_run = (
        final_weekly_transfer.groupby(
            ["file", "run_id", "ward_level", "moving_average_rounded"]
        )
        .agg(
            weekly_transfer=("weekly_transfer", "mean")
        )  # or sum, if that makes more sense
        .reset_index()
    )

    # Step 2: Then aggregate across runs for confidence interval
    aggregated_data = (
        per_run.groupby(["file", "ward_level", "moving_average_rounded"])
        .agg(
            weekly_transfer_mean=("weekly_transfer", np.mean),
            weekly_transfer_ci_lower=(
                "weekly_transfer",
                lambda x: mean_confidence_interval(x)[1],
            ),
            weekly_transfer_ci_upper=(
                "weekly_transfer",
                lambda x: mean_confidence_interval(x)[2],
            ),
        )
        .reset_index()
    )

    return aggregated_data


filtered_data_frames = filter_data_frames(admissions)
final_weekly_transfer = merge_and_calculate(
    filtered_data_frames, occupancy_rates, warm_up_time=0
)
aggregated_data = calculate_aggregated_data(final_weekly_transfer)
plot_weekly_transfer(aggregated_data)
```

## Individual Ward Graphs with Error Bars (No Overlap)

Creating separate graphs for each ward level with error bars instead of confidence bands to eliminate visual overlap while maintaining the exact same styling and comparisons.


```python
def plot_weekly_transfer_individual_wards_errorbar(aggregated_data):
    """
    Plot individual graphs for each ward level with error bars instead of confidence bands.
    Maintains exact same styling as original plots but eliminates visual overlap.
    """
    # Define labels for each scenario (exactly as original)
    labels = ["Base", "Sepsis", "Phototherapy", "Change NICU guidelines", "Combined"]

    # Mapping for better titles (exactly as original)
    ward_title_map = {"NICU": "NICU", "high": "HCN", "medium": "MCN"}

    # Unique ward levels
    ward_levels = aggregated_data["ward_level"].unique()

    # Create a color palette and map for files (exactly as original)
    palette = sns.color_palette("tab10", aggregated_data["file"].nunique())
    color_map = {
        file_num: color
        for file_num, color in zip(aggregated_data["file"].unique(), palette)
    }

    # Plot separate graphs for each ward level
    for ward_level in ward_levels:
        subset = aggregated_data[aggregated_data["ward_level"] == ward_level]
        plt.figure(figsize=(10, 6), dpi=400)

        for file_num in subset["file"].unique():
            file_subset = subset[subset["file"] == file_num]

            # Calculate error bars (difference from mean to CI bounds)
            error_lower = (
                file_subset["weekly_transfer_mean"]
                - file_subset["weekly_transfer_ci_lower"]
            )
            error_upper = (
                file_subset["weekly_transfer_ci_upper"]
                - file_subset["weekly_transfer_mean"]
            )

            # Plot with error bars instead of confidence bands
            plt.errorbar(
                file_subset["moving_average_rounded"],
                file_subset["weekly_transfer_mean"],
                yerr=[error_lower, error_upper],
                marker="o",
                markersize=3,
                label=labels[file_num],
                color=color_map[file_num],
                linewidth=1,
                capsize=3,
                capthick=1,
            )

        # Exact same styling as original
        plt.title(f"{ward_title_map.get(ward_level, ward_level)} Ward Level")
        plt.xlabel("Weekly Occupancy (%)")
        plt.ylabel("Average Weekly Rejections")
        plt.legend()
        plt.grid(True, linestyle="--", alpha=0.7)
        plt.tight_layout()
        plt.show()


# Execute the function
plot_weekly_transfer_individual_wards_errorbar(aggregated_data)
```
